#  Исследование результатов А/В-теста и поиск инсайтов

SollmaFin — международное мобильное приложение для онлайн-торговли финансовыми активами. Пользователи могут инвестировать деньги в акции, валюту, криптовалюту, биржевые фонды (ETF) и другие активы. Целевая аудитория приложения — начинающие инвесторы. Получить доступ к финансовым рынкам можно со смартфона либо через веб-версию. 

Приложение ориентировано на рынок Латинской Америки с акцентом на четыре страны: Мексику, Бразилию, Колумбию и Аргентину — там быстро растёт интерес к финтеху и инвестициям, а экономики очень разные. 

Появилась гипотеза о недостаточной финансовой грамотности пользователей: они не понимают, как работать с активами разной степени риска. Клиенты покупают активы с высоким риском и теряют деньги, что снижает вовлечённость.
Команда продукта решила обновить онбординг пользователей и добавить в него детальную информацию о различиях финансовых активов и связанных с ними рисках. Однако появилось опасение, что углублённый онбординг отпугнёт клиентов от пополнения депозита и более рискованных вложений. 

Чтобы разобраться, команда решила провести A/B-эксперимент. 

### А/В-эксперимент

В эксперименте участвовали новые пользователи, которые зарегистрировались в приложении со 2 по 15 июня 2025 года. 
Пользователей случайным образом разделили на две равные группы: 
- Контрольная группа проходила стандартный онбординг без обязательного обучения.
- Тестовая группа проходила обновлённый онбординг с подробной информацией об активах и связанных с ними рисках. 

После разделения активность пользователей анализировали в течение недели.

Цель эксперимента — оценить, как обновлённый онбординг влияет на поведение пользователей и их дальнейшую инвестиционную активность.

При внедрении новой фичи команда продукта выдвинула такие гипотезы:
- Гипотеза роста: обучающий онбординг помогает пользователям лучше понимать принципы инвестирования, поэтому они будут чаще открывать второй депозит.
- Гипотеза риска: информация о возможных потерях и высоких рисках отпугнёт некоторых новичков, особенно самых осторожных, что снизит конверсию в первый депозит.
- Дополнительная гипотеза: после нового онбординга пользователи, которые выбрали высокорискованные активы, будут чаще, чем раньше, возвращаться и открывать второй депозит. При старом онбординге пользователи часто покупали активы с высоким риском без понимания последствий. Это приводило к потерям и оттоку после первого депозита. 

Чтобы увидеть полную картину, недостаточно следить за метриками конверсии. Команда выбрала для эксперимента такой набор метрик: 
- Ключевая метрика — средняя сумма всех депозитов на одного пользователя (включая тех, кто установил приложение или открыл веб-версию).
- Барьерная метрика — конверсия из регистрации в первый депозит.
- Вспомогательная метрика 1 — конверсия из первого депозита во второй.
- Вспомогательная метрика 2 — средняя сумма всех депозитов на пользователя, который открыл хотя бы один депозит.

Ожидалось, что ключевая и барьерная метрики не упадут, а вспомогательные покажут значительный рост.

**Задача 2.1.** Анализ новых пользователей.
- Изучите динамику привлечения новых пользователей в приложение.

````python
#Приводим к неделям
history['first_week'] = history['first_dt'].dt.to_period('W').dt.to_timestamp()

#Создаем агрегат уникальных пользователей по неделям
df_agg = history.groupby('first_week')['user_id'].nunique().reset_index()

# Строим линейный график
plt.figure(figsize=(12,6))
plt.plot(df_agg['first_week'], df_agg['user_id'], color='skyblue')
plt.title('Динамика привлечения новых пользователей')
plt.xlabel('Дата')
plt.ylabel('Количество уникальных пользователей')
plt.tight_layout()
plt.grid(axis='y')

plt.show()
````

````python
# Выбор признака сегментации
def plot_segments(data: pd.DataFrame, segmentation_attribute: str) -> None:
    df_gp = data.groupby(["first_week", segmentation_attribute], as_index=False)["user_id"].nunique()
    
    # Получение всех уникальных сегментов
    segments = df_gp[segmentation_attribute].unique()

    plt.figure(figsize=(12, 6))
    # Для каждого уникального сегмента строим линейный график
    for segment in segments:
       df_plot = df_gp[df_gp[segmentation_attribute] == segment]
       plt.plot(df_plot['first_week'], df_plot['user_id'], label=segment)
       
    # Устанавливаем заголовок
    plt.title(f"Сегментация по признаку {segmentation_attribute}")

    # Устанавливаем подписи осей
    plt.xlabel("Неделя")
    plt.ylabel("WAU")
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.grid(axis='y')

    # Добавляем легенду
    plt.legend()
    plt.show()


# Вызов функции для признака страны
plot_segments(history, 'country_code')

# Вызов функции по платформам
plot_segments(history, 'platform')
````

Динамика привлечения пользователей стабильна и без резких спадов и роста.

По странам есть небольшое снижение в Мексике и Аргентине.
По платформам основная аудитория была с мобильных устройств. Динамика стабильна.

**Задача 2.2.** Анализ воронок событий.
- Постройте классическую и пошаговую воронки;
- Охарактеризуйте пользовательский путь в приложении и определите узкие места воронки. 


````python
#Создаем список воронки
funnel_steps = ['install / open_web', 'registration', 'main_page', 'onboarding_complete',
'first_deposit','asset_purchase', 'second_deposit']

#Фильтруем по шагам воронки
funnel_df = history[history['event_name'].isin(funnel_steps)]
#Считаем количество пользователей на каждом этапе
funnel_df = funnel_df.groupby('event_name')['user_id'].nunique().reindex(funnel_steps).reset_index().rename(columns={'user_id':'users_n'})
#Определеяем количество пользователей на первом этапе
first_step_users = funnel_df.iloc[0]['users_n']
#Cчитаем конверсию
funnel_df['cr'] = (funnel_df['users_n'] / first_step_users) * 100

display(funnel_df)

#Визуализируем данные в график
plt.figure(figsize=(12,6))

bars = plt.bar(
    funnel_df['event_name'],
    funnel_df['cr'],
    color='skyblue'
)

# Добавляем подписи над столбцами
plt.bar_label(
    bars,
    fmt='%.1f%%',
    padding=3
)

plt.title('Классическая воронка')
plt.xlabel('Этап воронки')
plt.ylabel('Конверсия, %')
plt.tight_layout()
plt.grid(axis='y')

plt.show()
````

Основная потеря аудитории происходит на этапе перехода к покупке актива, на котором конверсия снижается на 30 п.п относительно предыдущего шага.
Также около 10 000 уникальных пользователей не дошли до второго депозита после покупки актива (конверсия снизилась на 20 п.п)

````python
#Создаем визуализацию step_by_step ворокни

#Создаем столбец с конверсией предыдущего шага
funnel_df['conversions_from_prev_step_%'] = funnel_df['users_n'] / funnel_df['users_n'].shift(1) * 100
display(funnel_df)
#Визуализируем данные в график
plt.figure(figsize=(12,6))
plt.bar(funnel_df['event_name'], funnel_df['conversions_from_prev_step_%'], color='skyblue')
plt.title('Step_by_step воронка')
plt.xlabel('Конверсия от предыдущего этапа')
plt.tight_layout()
plt.grid(axis='y')
plt.show()
````

Step-by-step также показывает, что на этапе покупки актива конверсия от предыдущего шага составляла 55%. Минимальная конверсия от предыдущего шага была на этапе второго депозита и составляла 35%.

**Задача 2.3.** Влияние уровня риска актива на открытие второго депозита.

````python
#Отфильтруем и отсортируем покупки. Оставим только уникальных пользователей 
df_first_purchase = history[history['event_name'] == 'asset_purchase'].sort_values(by='event_ts', ascending=True).drop_duplicates(subset=['user_id'], keep='first', inplace=False)
#Отфильтруем уникальных пользователей с вторым депозитом
df_second_deposit = history[history['event_name'] == 'second_deposit']['user_id'].unique()

#Создадим столбец где укажем был ли у пользователей с покупкой второй депозит
df_first_purchase['has_second_deposit'] = df_first_purchase['user_id'].isin(df_second_deposit)

#Агрегируем по уровню риска и посчитаем среднюю конверсию во второй депозит
df_first_purchase_agg = (df_first_purchase.groupby('risk_level')['has_second_deposit'].mean()*100).reset_index().rename(columns={'has_second_deposit':'cr_to_second_dep,%'})
display(df_first_purchase_agg)
````

Активы с высоким риском показывали минимальную конверсию во второй депозит и составляла 26%. Конверсия во второй депозит у активов с низким и средним риском составляла 46% и 42% соответственно.

````python
# Создаем пустой список, куда сложим данные для каждой группы
result_funnels = []

# Циклом проходимся по каждой уникальной группе
for group_name in df_ab['group'].unique():
    
    #Делим по группам
    df_ab_group = df_ab[df_ab['group'] == group_name]
    
    #Учитываем только события из воронки
    funnel_df_ab = df_ab_group[df_ab_group['event_name'].isin(funnel_steps)]
    #Считаем количество уникальных пользователей на каждом этапе
    funnel_df_ab = funnel_df_ab.groupby('event_name')['user_id'].nunique().reindex(funnel_steps).reset_index().rename(columns={'user_id':'users_n'})
    
    #Считаем количество пользователей на первом шаге
    first_step_users = funnel_df_ab.iloc[0]['users_n']
    #Считаем конверсию по классической воронке
    funnel_df_ab['cr'] = (funnel_df_ab['users_n'] / first_step_users) * 100
    #Считаем конверсию для step-by-step воронки
    funnel_df_ab['conversions_from_prev_step_%'] = funnel_df_ab['users_n'] / funnel_df_ab['users_n'].shift(1) * 100

    #Добавляем колонку с именем группы
    funnel_df_ab['group'] = group_name
    
    #Сохраняем в список
    result_funnels.append(funnel_df_ab)

# Объединяем список из таблиц и выводим результат
funnel_df_ab = pd.concat(result_funnels)
display(funnel_df_ab)
````

**Задача 3.3.** Влияние новой фичи на конверсию во второй депозит с учётом уровня риска купленного актива.
- Проанализируйте, как категория риска купленного актива влияет на вероятность открытия второго депозита в тестовой и контрольной группах. 
- Сравните результаты и оцените, есть ли статистически значимая разница между группами. 

````python
#Оставляем только первую покупки каждого пользователя
df_ab_first_purchase = df_ab[df_ab['event_name'] == 'asset_purchase'].sort_values(by='event_ts', ascending=True).drop_duplicates(subset=['user_id'], keep='first', inplace=False) 
#Оставляем вторые депозиты на уникального пользователя
df_ab_second_deposit = df_ab[df_ab['event_name'] == 'second_deposit']['user_id'].unique()

#Создаем столбец, где укажем был ли второй депозит у тех, кто совершил покупку актива
df_ab_first_purchase['has_second_deposit'] = df_ab_first_purchase['user_id'].isin(df_ab_second_deposit)

#Создаем агрегат по группе теста и уровню риска. Рассчитываем CR, количество пользователей и сумму депозитов
df_exp = df_ab_first_purchase.groupby(['group', 'risk_level']).agg(
    conversion_rate=('has_second_deposit', 'mean'), 
    unique_users=('user_id', 'nunique'),         
    total_deposits=('has_second_deposit', 'sum') 
).reset_index()

#Проверим достаточно ли у нас данных для z-теста пропорций
# 1. Предварительные вычисления (успехи и неудачи)
df_exp['successes'] = df_exp['total_deposits']
df_exp['failures'] = df_exp['unique_users'] - df_exp['total_deposits']

# 2. Проверяем условие для всех строк
condition_success = (df_exp['successes'] > 10).all()
condition_failures = (df_exp['failures'] > 10).all()

# 3. Выводим результат
if condition_success and condition_failures:
    print('Предпосылка о достаточном количестве данных выполняется для всех сегментов!')
else:
    print('Предпосылка о достаточном количестве данных НЕ выполняется в некоторых сегментах!')
````

Проверим гипотезу о влияние фичи на конверсию с учетом уровня риска купленного актива.

H0: Новая фича не влияет на конверсию во второй депозит по разным уровням риска купленного актива 

H1: Новая фича положительно сказывается на конверсию во второй депозит в разрезе риска купленного актива

````python
#Проверим стат значимость
from statsmodels.stats.proportion import proportions_ztest

#Учтем поправку на множественное тестирование
alpha = 0.05 / 3

#Сохраняем в переменную уровни риска
risk_levels = df_exp['risk_level'].unique()

#Создаем цикл по уровням риска
for risk in risk_levels:
    print(f'H0: Новая фича не влияет на конверсию во второй депозит для активов уровня риска - {risk}\nH1: Новая фича влияет на конверсию во второй депозит для активов уровня риска - {risk}')
    
    # Берём из датафрейма размеры выборок
    n_a = df_exp[(df_exp['group'] == 'control') & (df_exp['risk_level'] == risk)]['unique_users'].iloc[0]
    n_b = df_exp[(df_exp['group'] == 'test') & (df_exp['risk_level'] == risk)]['unique_users'].iloc[0]
    
    # Берём количество успехов
    m_a = df_exp[(df_exp['group'] == 'control') & (df_exp['risk_level'] == risk)]['total_deposits'].iloc[0]
    m_b = df_exp[(df_exp['group'] == 'test') & (df_exp['risk_level'] == risk)]['total_deposits'].iloc[0]
    
    print(f'Control: конверсия {m_a}/{n_a} ({m_a/n_a*100:.2f}%)')
    print(f'Test: конверсия {m_b}/{n_b} ({m_b/n_b*100:.2f}%)')
    
    # Проводим двусторонний тест
    stat_ztest, p_value_ztest = proportions_ztest(
    [m_b, m_a],  # test, control
    [n_b, n_a],
    alternative='larger')

    
    if p_value_ztest > alpha:
        print(f'p-value = {p_value_ztest:.4f} > {alpha}')
        print('Нет оснований отвергнуть нулевую гипотезу.')
    else:
        print(f'p-value = {p_value_ztest:.4f} <= {alpha}')
        print('Нулевая гипотеза отвергается.')
    print()
````

На основе проведения теста есть основания отвергнуть нулевую гипотезу для активов с высоким уровнем риска. Наблюдали статистически значимый рост конверсии во второй депозит в тестовой группе для высокорисковых активов. Для низко- и среднерисковых активов значимой разницы между группами не обнаружено.

````python
from scipy import stats

#Создаем агрегат для расчета суммы выручки на каждого пользователя в группах теста
user_deposits = df_ab.groupby(['user_id', 'group'], as_index=False)['amount'].sum()

#Разделяем данные по группам
control_data = user_deposits[user_deposits['group'] == 'control']['amount']
test_data = user_deposits[user_deposits['group'] == 'test']['amount']

# Считаем средние
mean_control = control_data.mean()
mean_test = test_data.mean()

#Проводм t-тест Уэлча
stat, p_value = stats.ttest_ind(test_data, control_data, equal_var=False)

#Считаем абсолютные и относительные разницы
abs_diff = mean_test - mean_control
rel_diff = (mean_test / mean_control) - 1

#Собираем результат в датафрейм и выводим
results_kpi = pd.DataFrame([{
    'Group A': 'control',
    'Group B': 'test',
    'Mean A': round(mean_control, 2),
    'Mean B': round(mean_test, 2),
    'p-value': p_value,
    'Absolute Lift': round(abs_diff, 2),
    'Relative Lift (%)': round(rel_diff * 100, 2)
}])

display(results_kpi)
````

По вспомогательным метрикам обнаружены статически значимые изменения между группами.
Есть основания отклонить нулевые гипотезы по вспомогательным метрикам и принять альтернативные гипотезы о наличии влияния фичи на конверсию из первого депозита во второй и увеличения суммы депозитов пользователей хотя бы с одним депозитом.


````python
df_ab['activity_date'] = pd.to_datetime(df_ab['event_ts']).dt.date
dep_events = df_ab[df_ab['event_name'].isin(['first_deposit', 'second_deposit'])]

daily_user_sums = dep_events.groupby(['user_id', 'activity_date', 'group'])['amount'].sum().reset_index()
daily_user_sums = daily_user_sums.rename(columns={'amount': 'daily_amount'})

all_users = df_ab[['user_id', 'group']].drop_duplicates()

records = []

for current_date in sorted(daily_user_sums['activity_date'].unique()):
    cum_sums = daily_user_sums[daily_user_sums['activity_date'] <= current_date]
    total = cum_sums.groupby(['user_id', 'group'])['daily_amount'].sum().reset_index()
    
    users_full = total.merge(all_users, on=['user_id', 'group'], how='right')
    users_full['daily_amount'] = users_full['daily_amount'].fillna(0)
    
    means = users_full.groupby('group')['daily_amount'].mean().reset_index()
    means.columns = ['group', 'mean_deposit']
    
    control_vals = users_full[users_full['group'] == 'control']['daily_amount']
    test_vals = users_full[users_full['group'] == 'test']['daily_amount']
    
    stat, p_value = stats.ttest_ind(test_vals, control_vals, equal_var=False)
    
    records.append({
        'date': current_date,
        'control_mean': means[means['group'] == 'control']['mean_deposit'].values[0],
        'test_mean': means[means['group'] == 'test']['mean_deposit'].values[0],
        'p_value': p_value
    })

cumulative_df = pd.DataFrame(records)

#Строим график по кумулятивным метрикам
plt.figure(figsize=(14,5))

plt.subplot(1,2,1)
plt.plot(cumulative_df['date'], cumulative_df['control_mean'], label='Control', marker='o')
plt.plot(cumulative_df['date'], cumulative_df['test_mean'], label='Test', marker='o')
plt.xticks(rotation=45)
plt.legend()
plt.title('Накопленное среднее суммы депозитов на пользователя')
plt.grid(True)

#Строим график p-value
plt.subplot(1,2,2)
plt.plot(cumulative_df['date'], cumulative_df['p_value'], color='red', marker='o')
plt.axhline(0.05, linestyle='--', color='gray', label='α=0.05')
plt.xticks(rotation=45)
plt.legend()
plt.title('Стабильность p-value')
plt.grid(True)

plt.tight_layout()
plt.show()
````

По динамике изменений ключевой метрики только с 10 июня показатель в тестовой группе стал стабильно выше метрики в контрольной группе.
p.value в течение всего эксперимента (кроме 2 июня) был выше выбранного уровня значимости.

````python
#Заводим цикл бутстрапа
for i in range(n_iterations):
    boot_control = np.random.choice(payers_control, size=len(payers_control), replace=True)
    boot_test = np.random.choice(payers_test, size=len(payers_test), replace=True)

    # Считаем три перцентиля для контрольной и тестовой выборок
    control_25 = np.percentile(boot_control, 25)
    test_25 = np.percentile(boot_test, 25)
    
    control_50 = np.percentile(boot_control, 50)
    test_50 = np.percentile(boot_test, 50)
    
    control_75 = np.percentile(boot_control, 75)
    test_75 = np.percentile(boot_test, 75)

    # Разницы и сохраняем в соответствующие списки
    diffs_25.append(test_25 - control_25)
    diffs_50.append(test_50 - control_50)
    diffs_75.append(test_75 - control_75)
````
